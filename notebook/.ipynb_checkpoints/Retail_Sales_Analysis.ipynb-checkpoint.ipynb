```python
# correcting dtype 
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
print(df['Order Date'].dtype)
print(df['Ship Date'].dtype)
```

```python
df['Order Date'].head(10)
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
```

```python
df[['Order Date']].head()
```

```python
print("Total Revenue:", round(df['Sales'].sum(), 2))
print("Total Profit:", round(df['Profit'].sum(), 2))
print("Total Orders:", df['Order ID'].nunique())
print("Total Customers:", df['Customer ID'].nunique())
```

```python
#Which category generates the most revenue?

category_sales = (
    df.groupby('Category')['Sales']
      .sum()
      .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x='Category',
    y='Sales',
    data=df,
    estimator=sum
)

plt.title('Revenue by Category')
plt.savefig("../image/revenuenycategory.png")
plt.show()
```

```python
# Business Interpretation
'''
Technology
Highest revenue-generating category.
Contributes roughly 36% of total sales.
Likely contains high-value products such as printers, machines, phones, and accessories.
Furniture
Second-largest contributor.
Strong revenue but often (we'll verify later) has lower margins due to discounts and shipping costs.
Office Supplies
Lowest revenue among the three.
However, it often generates consistent profits because of frequent purchases.
'''
```

```python
#Top 10 Products by Revenue
top_products = (
    df.groupby('Product Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
top_products
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title('Top 10 Products by Revenue')
plt.xlabel('Revenue')
plt.ylabel('Product')
plt.savefig("../image/top10.png")
plt.show()
```

```python
#Observations
'''
The highest revenue product (Canon imageCLASS 2200 Advanced Copier) generated $61,599, which is more than double the second-ranked product.
Most top-selling products belong to the Technology category.
High-ticket items (copiers, printers, videoconferencing systems) are driving a significant portion of revenue.
'''
```

```python
#revenue by region

region_sales = (
    df.groupby('Region')['Sales']
      .sum()
      .sort_values(ascending=False)
)
print(region_sales)
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title('Revenue by Region')
plt.xlabel('Region')
plt.ylabel('Revenue')
plt.savefig("../image/regional")
plt.show()
```

```python
#profit by region 

region_profit = (
    df.groupby('Region')['Profit']
      .sum()
      .sort_values(ascending=False)
)

print(region_profit)
```

```python
## Monthly Sales Trends
monthly_sales = (
    df.groupby('Month')['Sales']
      .sum()
      .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales['Month'] = monthly_sales['Month'].astype(str)
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_sales['Month'],
    monthly_sales['Sales']
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')

plt.xticks(rotation=90)

plt.show()
```

```python
# Profit Analysis

category_profit = (
    df.groupby('Category')['Profit']
      .sum()
      .sort_values(ascending=False)
)

category_profit
plt.figure(figsize=(8,5))
sns.barplot(
    x=category_profit.index,
    y=category_profit.values
)
plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.savefig("../image/profitanalysis")

plt.show()
```
